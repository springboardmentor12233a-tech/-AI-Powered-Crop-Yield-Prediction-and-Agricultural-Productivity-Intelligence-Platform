```python
import os
import subprocess

# 1. Convert notebook to HTML
subprocess.run(
    ['python', '-m', 'jupyter', 'nbconvert', '--to', 'html', 'milestone1.ipynb']
)

# 2. Locate Microsoft Edge
edge_paths = [
    r'C:\Program Files (x86)\Microsoft\Edge\Application\msedge.exe',
    r'C:\Program Files\Microsoft\Edge\Application\msedge.exe',
]
edge_exe = next((p for p in edge_paths if os.path.exists(p)), None)

if edge_exe:
  html_file = os.path.abspath('milestone1.html')
  pdf_file = os.path.abspath('milestone1.pdf')

  cmd = [
      edge_exe,
      '--headless',
      '--disable-gpu',
      f'--print-to-pdf={pdf_file}',
      '--no-pdf-header-footer',
      html_file,
  ]
  subprocess.run(cmd)
  print('PDF generated successfully at:', pdf_file)
else:
  print('Edge executable not found. Open milestone1.html in your browser and press Ctrl+P to save as PDF.')
```
