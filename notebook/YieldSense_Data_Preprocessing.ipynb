```python
import pandas as pd

# Load cleaned dataset
train = pd.read_csv("cleaned_crop_yield_train.csv")
test = pd.read_csv("cleaned_crop_yield_test.csv")

# ============================================
# 1. FIND CROP FREQUENCY
# ============================================

crop_counts = train["Crop"].value_counts()

print("Original number of crops:", train["Crop"].nunique())

print("\nCrop counts:")
print(crop_counts)


# ============================================
# 2. KEEP ONLY CROPS WITH >= 50 RECORDS
# ============================================

valid_crops = crop_counts[crop_counts >= 50].index

train = train[train["Crop"].isin(valid_crops)]
test = test[test["Crop"].isin(valid_crops)]


# ============================================
# 3. DISPLAY REMOVED CROPS
# ============================================

removed_crops = crop_counts[crop_counts < 50]

print("\nRemoved crops:")
print(removed_crops)


# ============================================
# 4. FINAL DATASET INFORMATION
# ============================================

print("\n============================================")
print("FINAL CLEANING SUMMARY")
print("============================================")

print("Training rows:", len(train))
print("Test rows:", len(test))

print("Number of crops:", train["Crop"].nunique())
print("Number of states:", train["State"].nunique())
print("Number of seasons:", train["Season"].nunique())

print("Year range:", train["Year"].min(), "-", train["Year"].max())

print("\nMissing values:")
print(train.isnull().sum())

print("\nDuplicate rows:", train.duplicated().sum())


# ============================================
# 5. SAVE FINAL CLEANED DATA
# ============================================

train.to_csv(
    "final_cleaned_crop_yield_train.csv",
    index=False
)

test.to_csv(
    "final_cleaned_crop_yield_test.csv",
    index=False
)

print("\nFinal cleaned datasets saved successfully!")
```

```python
import pandas as pd

train = pd.read_csv("../dataset/cleaned_crop_yield_train.csv")
test = pd.read_csv("../dataset/cleaned_crop_yield_test.csv")

# Count crops in training data
crop_counts = train["Crop"].value_counts()

# Keep crops with at least 50 records
valid_crops = crop_counts[crop_counts >= 50].index

# Apply the same crop list to both datasets
final_train = train[train["Crop"].isin(valid_crops)].copy()
final_test = test[test["Crop"].isin(valid_crops)].copy()

# Show removed crops
print("Removed crops:")
print(crop_counts[crop_counts < 50])

print("\nFinal train shape:", final_train.shape)
print("Final test shape:", final_test.shape)

print("\nRemaining crops:", final_train["Crop"].nunique())

# Save
final_train.to_csv(
    "../dataset/final_cleaned_crop_yield_train.csv",
    index=False
)

final_test.to_csv(
    "../dataset/final_cleaned_crop_yield_test.csv",
    index=False
)

print("\nFinal cleaned datasets saved successfully!")
```
