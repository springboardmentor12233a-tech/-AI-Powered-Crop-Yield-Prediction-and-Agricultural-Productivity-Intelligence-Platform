```python
import pandas as pd

original_train = pd.read_csv(
    r"C:\Users\NS\Desktop\YieldSense-AI\dataset\crop_yield_train.csv"
)

original_train["Calculated_Yield"] = (
    original_train["Production"] / original_train["Area"]
)

print(original_train[
    ["Production", "Area", "Yield", "Calculated_Yield"]
].head(10))

difference = (
    original_train["Yield"] -
    original_train["Calculated_Yield"]
).abs()

print("\nMaximum difference:", difference.max())
print("Mean difference:", difference.mean())
print("Median difference:", difference.median())
```
