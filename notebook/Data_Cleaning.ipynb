```python
import pandas as pd

crop_df = pd.read_csv("../dataset/Crop_Yield_Test.csv")
faostat_df = pd.read_csv("../dataset/FAOSTAT_Crop_Production.csv")
```

```python
crop_df["Year"] = pd.to_datetime(crop_df["harvest_date"]).dt.year
```

```python
crop_df["Year"].value_counts()
```

```python
faostat_df = faostat_df[faostat_df["Area"] == "India"]
```

```python
faostat_wide = faostat_df.pivot_table(
    index=["Item", "Year"],
    columns="Element",
    values="Value",
    aggfunc="first"
).reset_index()
```

```python
faostat_wide.head()
```

```python
faostat_wide["Item"].unique()
```

```python
crop_df["crop_type"].unique()
```

```python
faostat_wide[
    faostat_wide["Item"].str.contains(
        "Rice|Soybean|Soybeans|Maize",
        case=False,
        na=False
    )
]["Item"].unique()
```

```python
faostat_wide[
    faostat_wide["Item"].str.contains(
        "soy|soya",
        case=False,
        na=False
    )
]["Item"].unique()
```

```python
crop_mapping = {
    "Rice": "Rice",
    "Soybean": "Soya beans",
    "Wheat": "Wheat",
    "Barley": "Barley",
    "Corn": "Maize (corn)"
}

crop_df["Item"] = crop_df["crop_type"].map(crop_mapping)
```

```python
combined_df = pd.merge(
    crop_df,
    faostat_wide,
    on=["Item", "Year"],
    how="left"
)
```

```python
combined_df["harvest_date"] = pd.to_datetime(combined_df["harvest_date"])
combined_df.info()
```

```python
combined_df = combined_df.rename(columns={
    "Area harvested": "area_harvested",
    "Production": "production",
    "Yield": "yield"
})
combined_df.columns
```

```python
combined_df.info()
```

```python
# Data Cleaning

# 1. Remove duplicate rows
combined_df = combined_df.drop_duplicates()

# 2. Convert harvest_date to datetime
combined_df["harvest_date"] = pd.to_datetime(
    combined_df["harvest_date"],
    errors="coerce"
)

# 3. Remove rows with invalid dates
combined_df = combined_df.dropna(subset=["harvest_date"])

# 4. Standardize categorical text
combined_df["crop_type"] = combined_df["crop_type"].str.strip()
combined_df["region"] = combined_df["region"].str.strip()
combined_df["season"] = combined_df["season"].str.strip()
combined_df["Item"] = combined_df["Item"].str.strip()
combined_df["field_id"] = combined_df["field_id"].str.strip()

# 5. Reset index after cleaning
combined_df = combined_df.reset_index(drop=True)

# 6. Display final shape
print("Cleaned dataset shape:", combined_df.shape)

# 7. Check missing values after cleaning
print("\nMissing values:")
print(combined_df.isnull().sum())

# 8. Check duplicates after cleaning
print("\nDuplicate rows:", combined_df.duplicated().sum())
```

```python
combined_df.to_csv("../dataset/Cleaned_Crop_Yield.csv", index=False)
```
