```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("../datasets/Smart_Farming_Crop_Yield_2024.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
# Handle missing categorical values
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")

print("Missing values after treatment:")
print(df.isnull().sum().sum())
```

```python
df["sowing_date"] = pd.to_datetime(df["sowing_date"], errors="coerce")
df["harvest_date"] = pd.to_datetime(df["harvest_date"], errors="coerce")

# Create useful date-based features
df["sowing_month"] = df["sowing_date"].dt.month
df["sowing_year"] = df["sowing_date"].dt.year

df.head()
```

```python
target = "yield_kg_per_hectare"

X = df.drop(columns=[target])
y = df[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
df.to_csv("../datasets/Smart_Farming_Crop_Yield_Preprocessed.csv", index=False)

print("Preprocessed dataset saved successfully!")
```

```python
# Remove ID columns
df = df.drop(columns=["farm_id", "sensor_id"], errors="ignore")

# Process timestamp
df["timestamp"] = pd.to_datetime(df["timestamp"], errors="coerce")

df["timestamp_month"] = df["timestamp"].dt.month
df["timestamp_hour"] = df["timestamp"].dt.hour

# Remove original date/timestamp columns
df = df.drop(columns=["sowing_date", "harvest_date", "timestamp"], errors="ignore")

print("Updated Shape:", df.shape)
df.head()
```

```python
target = "yield_kg_per_hectare"

X = df.drop(columns=[target])
y = df[target]

print("X Shape:", X.shape)
print("y Shape:", y.shape)
print("\nCategorical Columns:")
print(X.select_dtypes(include="object").columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

categorical_features = X_train.select_dtypes(include="object").columns.tolist()
numerical_features = X_train.select_dtypes(exclude="object").columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_features)
    ]
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed Training Shape:", X_train_processed.shape)
print("Processed Testing Shape:", X_test_processed.shape)
```

```python
print("Preprocessing completed successfully!")
print("Training samples:", len(X_train_processed))
print("Testing samples:", len(X_test_processed))
print("Target training samples:", len(y_train))
print("Target testing samples:", len(y_test))
```
