```python
#correlation analysis
# Keep the original dataset unchanged
df_clean = data.copy()  
numeric_columns = df_clean.select_dtypes(include=["number"]).columns
correlation_matrix = df_clean[numeric_columns].corr()
correlation_matrix
#visualize the correlation matrix
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
#specific correlation with yield
yield_correlation = correlation_matrix["yield_tpha"].sort_values(ascending=False)
print("Correlation of features with yield (kg per hectare):")
print(yield_correlation)
```

```python
#strong positive correlation with yield: soil_ph, total_rainfall, avg_temperature, fertilizer_amount, soil_moisture
yield_corr=(yield_correlation[yield_correlation > 0.5])
print("Features with strong positive correlation (|r| > 0.5) with yield:")
print(yield_corr)
```
