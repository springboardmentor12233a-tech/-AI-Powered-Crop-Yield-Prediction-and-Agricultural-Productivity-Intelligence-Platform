```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/Crop_recommendation.csv')
print("Data loaded successfully! Here are the first 5 rows:")
df.head()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x="label", y="rainfall", palette="viridis")
plt.title("Average Rainfall Requirement by Crop Type")
plt.xlabel("Crop Type")
plt.ylabel("Average Rainfall (mm)")
plt.xticks(rotation=45) # Rotates labels so they don't overlap
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="rainfall", y="temperature", hue="label", alpha=0.7)
plt.title("Rainfall vs Temperature by Crop Type")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Temperature (°C)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))
numeric_cols = df.select_dtypes(include=['float64', 'int64'])
correlation_matrix = numeric_cols.corr()

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Environmental Factors")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.histplot(data=df, x="temperature", kde=True, ax=axes[0], color="orange")
axes[0].set_title("Distribution of Temperature")

sns.countplot(data=df, x="label", ax=axes[1])
axes[1].set_title("Number of Records by Crop Type")
axes[1].tick_params(axis='x', rotation=90) # Rotates labels to prevent overlap

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(data=df, x="label", y="N", ax=axes[0])
axes[0].set_title("Nitrogen (N) by Crop")
axes[0].tick_params(axis='x', rotation=90)

sns.boxplot(data=df, x="label", y="P", ax=axes[1])
axes[1].set_title("Phosphorus (P) by Crop")
axes[1].tick_params(axis='x', rotation=90)

sns.boxplot(data=df, x="label", y="K", ax=axes[2])
axes[2].set_title("Potassium (K) by Crop")
axes[2].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()
```
